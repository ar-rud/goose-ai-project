in this notebook we change goose data set structure to make it work with unet

```python
import os
import shutil
from pathlib import Path
```

```python


# Base directory for the raw GOOSE dataset
# Adjust relative path if raw dataset is placed inside repo (e.g., Path('../goose_dataset'))
RAW_DATASET_DIR = Path("../goose-dataset")

# Source directories (relative to RAW_DATASET_DIR)
img_src = RAW_DATASET_DIR / "images" / "train"
label_src = RAW_DATASET_DIR / "labels" / "train"

# Destination directories (relative to unet_baseline folder)
DEST_DIR = Path("./data_goose")
img_dst = DEST_DIR / "imgs"
label_dst = DEST_DIR / "masks"

# Ensure destination directories exist
img_dst.mkdir(parents=True, exist_ok=True)
label_dst.mkdir(parents=True, exist_ok=True)
os.makedirs(img_dst, exist_ok=True)
os.makedirs(label_dst, exist_ok=True)

img_count = 0
for root, _, files in os.walk(img_src):
    for file in files:
        if file.endswith("_vis.png"):
            new_name = file.replace("_windshield_vis.png", ".png")
            shutil.copy(os.path.join(root, file), os.path.join(img_dst, new_name))
            img_count += 1

label_count = 0
for root, _, files in os.walk(label_src):
    for file in files:
        if file.endswith("_labelids.png"):
            new_name = file.replace("_labelids.png", ".png")
            shutil.copy(os.path.join(root, file), os.path.join(label_dst, new_name))
            label_count += 1

print(f"Copied {img_count} images and {label_count} masks.")
if img_count == label_count and img_count > 0:
    print("Success: 1:1 match between images and masks.")
else:
    print("Warning: Unequal number of images and masks found, or zero files copied!")
```

```python
img_dir =  img_dst
count = 0
for filename in os.listdir(img_dir):
    if "_windshield.png" in filename:
        old_path = os.path.join(img_dir, filename)
        new_path = os.path.join(img_dir, filename.replace("_windshield.png", ".png"))
        os.rename(old_path, new_path)
        count += 1

print(f"Renamed {count} images to match mask filenames.")
```
