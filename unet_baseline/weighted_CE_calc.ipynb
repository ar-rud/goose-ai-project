here coefficients for weighted cross entropy are calculated

```python
import os
import numpy as np
from PIL import Image
from tqdm import tqdm
```

```python


gt_masks_dir = "../data_goose/masks_12class"
num_classes = 12
class_counts = np.zeros(num_classes, dtype=np.int64)

print("Counting pixels across masks...")
for file in tqdm(os.listdir(gt_masks_dir)):
    if file.endswith(".png"):
        mask = np.array(Image.open(os.path.join(gt_masks_dir, file)))
        unique, counts = np.unique(mask, return_counts=True)
        for u, c in zip(unique, counts):
            if u < num_classes:
                class_counts[u] += c

# Calculate Inverse Frequency (Median Frequency Balancing)
total_pixels = class_counts.sum()
class_frequencies = class_counts / total_pixels
median_freq = np.median(class_frequencies)

# Formula: weight = median_frequency / class_frequency
weights = median_freq / (class_frequencies + 1e-8) # 1e-8 avoids division by zero
weights = np.clip(weights, 0.1, 50.0) # Prevent weights from exploding to 1000x

print("\nCopy-paste this array into train_pipeline.py:")
print(f"[{', '.join([f'{w:.4f}' for w in weights])}]")
```
